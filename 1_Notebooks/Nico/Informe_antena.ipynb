## Diseño de antenas:
En el siguiente cuaderno de Jupyter se analizarán los modelos de antenas y los resultados obtenidos por Nicolás Pereyra, con el objetivo de su futura utilización en el proyecto AP3. Se evaluarán los resultados correspondientes a diferentes configuraciones de antenas, todas derivadas del diseño clásico Bow-Tie y sus modificaciones, orientadas a lograr un mayor ancho de banda.

Las simulaciones fueron realizadas empleando el software Ansys Electronics.

## Bowtie recortada:
Siguiendo el planteamiento del artículo “A 900 MHz Shielded Bow-Tie Antenna System for Ground Penetrating Radar”, se diseñó una antena Bow-Tie simple, pero con recortes en los extremos, tal como se muestra en la imagen. Este recorte modifica tanto el ancho de banda como la frecuencia central de la antena.

El diseño inicial se obtuvo utilizando el asistente (wizard) de Ansys Electronics y posteriormente se realizaron ajustes manuales para alcanzar la frecuencia central deseada.

## Bowtie-Rounded con metamaterial y plano reflectante:
No se cuentan con simulaciones de este mismo modelo sin los aditamentos; sin embargo, su rendimiento sin los planos adicionales fue similar al observado previamente.

El objetivo de incorporar estos planos es mejorar el ancho de banda (ampliar la región en la que S11 es menor a -10 dB), incrementar la direccionalidad y aumentar la ganancia. Para alcanzar los valores finales, se realizaron varias mediciones modificando las distancias y formas tanto del plano reflectante como del metamaterial, basándonos en las referencias del artículo “Directional and High-Gain Ultra-Wideband Bow-Tie Antenna for Ground-Penetrating Radar Applications”.

El diseño de la antena Bow-Tie fue desarrollado utilizando el asistente (wizard) de Ansys Electronics.
